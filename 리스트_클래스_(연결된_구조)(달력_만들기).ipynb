```python
#달력제작

일 = ["일","월","화","수","목","금","토"]
월 = {'1월':'31','2월':'28','3월':'31','4월':'30','5월':'31',
     '6월':'30','7월':'31','8월':'31','9월':'30','10월':'31',
     '11월':'30','12월':'31'}
```

```python
class 달력제작:

    def __init__(self,일들, 월들):
        self.일 = 일들
        self.월 = 월들
        self.일년달력=[]

    def 달력생성(self,일1):
        요일위치= self.일.index(일1)

        for 월들 in self.월:
            총일수 = int(self.월[월들])

            for 일 in range(1, 총일수 +1):
                현재요일 = self.일[요일위치]
                날짜정보={"월": 월들 ,"일":일 ,"요일": 현재요일}
                self.일년달력.append(날짜정보)

                요일위치=요일위치+1
                if 요일위치>6:
                    요일위치 =0

    def 달력출력(self):
        print ("1년치 달력 완성")
        for 날짜 in self.일년달력:
            print(str(날짜["월"])+ " " + str(날짜["일"])
            + "일 "+str(날짜["요일"])+ "요일")
```

```python
달력 = 달력제작(일,월)
달력.달력생성("목")
달력.달력출력()
```
