```python
쟁반 = 0;

class 쟁반탑:
    def __init__(self):
        self.쟁반층 = []

    def 쟁반쌓기(self, 쟁반):
        쟁반번호 = len(self.쟁반층)
        이름표 = str(쟁반번호)+"번 "


        self.쟁반층.append(쟁반)
        현재개수 = len(self.쟁반층)
        print(이름표+ "쟁반" + str(현재개수) +"장이 쌓였습니다.")

    def 쟁반사용(self):
        if len(self.쟁반층) >0:
            self.쟁반층.pop()
            쟁반번호 = len(self.쟁반층)
            이름표 = str(쟁반번호)+"번 "
            현재개수 = len(self.쟁반층)
            print(이름표+ "쟁반 사용, 남은 쟁반은" + str(현재개수) +"남았습니다.")
        else:
            print("남은 쟁반이 없습니다.")

```

```python
쟁반층 = 쟁반탑()
쟁반층.쟁반쌓기("쟁반1")
쟁반층.쟁반쌓기("쟁반2")
쟁반층.쟁반쌓기("쟁반3")
쟁반층.쟁반사용()
쟁반층.쟁반사용()
쟁반층.쟁반사용()
쟁반층.쟁반사용()
```
